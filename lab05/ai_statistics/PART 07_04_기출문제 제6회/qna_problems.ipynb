# 문제 1. 
- K-S 검정을 통해 업무 수행 시간이 정규분포를 따르는지 검정하고자 한다.
- 직원들의 업무 수행 시간의 평균과 표준편차를 구하시오. (소수점 셋째 자리까지 반올림)

```python
import pandas as pd
data = pd.read_csv('6_3_1.csv')
```

```python
# 코드
mean = round(data['work_hours'].mean(),3)
std = round(data['work_hours'].std(),3)
print(f"직원들의 업무 수행 시간의 평균 : {mean}")
print(f"직원들의 업무 수행 시간의 표준편차 : {std}")
```

- 직원들의 업무 수행 시간이 정규분포를 따르는지 K-S 검정을 실시하고, 검정통계량을 계산하시오. (소수점 셋째 자리까지 반올림)
- 기본 가설설정
  + 귀무가설 : 주어진 표본은 00분포를 따른다.
  + 대립가설 : 주어진 표본은 00분포를 따르지 않는다.
- 응용 가설설정
  + 귀무가설 : 주어진 표폰은 정규분포를 따른다. (reference 분포 = 두가지 분포가 일치하는지)
  + 대립가설 : 주어진 표폰은 정규분포를 따르지 않는다.
- kstest(rvs,cdf,args)
  + rvs : 데이터(우리가 주는 분포)
  + cdf : 참조할 분포
  + args : mean , std

```python
from scipy.stats import kstest, norm
# 코드
statistic, p_value = kstest(data['work_hours'], 'norm', args= (mean,std))
statistic.round(3), p_value.round(3)
```

- p-value를 바탕으로 `유의수준 5%`에서 귀무가설의 기각/채택 여부를 결정하시오. (p-value는 소수점 셋째 자리까지 반올림)

```python
# 코드
p_value = p_value.round(3)
if p_value < 0.05:
    print("귀무가설 기각")
else:
    print("귀무가설 채택")
```

# 문제 2. 
- 주택 가격을 종속 변수로 하고, 면적, 방의 개수, 연식을 독립 변수로 하는 다중회귀 분석을 수행하여, 회귀계수가 가장 높은 변수를 구하시오. (다중회귀모형 적합 시 절편 포함)
- 유의수준 5% 하에서 각 독립 변수가 주택 가격에 미치는 영향이 통계적으로 유의미한지 판단하고 유의미한 변수 개수를 구하시오. 

```python
import pandas as pd
data = pd.read_csv('6_3_2.csv')
data.head(1)
```

```python
# 코드
# 독립변수 종속변수 설정
import statsmodels.api as sm

X = data[['area','rooms','age']]
y = data[['price']]

# 상수항 추가
X = sm.add_constant(X)

# 다중회귀 분석 모델
model = sm.OLS(y, X).fit()
print(model.summary())
```

```python
# 코드
coefficients = model.params[1:]

print('회귀계수가 가장 큰 변수 :', coefficients.idxmax())
```

```python
# 코드
import numpy as np
p_values = model.pvalues[1:]
np.sum(p_values < 0.05)
```
