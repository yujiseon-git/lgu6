# 조건부 확률의 개념
- 어떤 사건 A가 이미 일어났다는 조건 하에 사건 B가 일어날 확률
- 예시
  + 전체 학생 중, 60%가 수학을 좋아함, P(A) = 0.6 (수학을 좋아하지 않는 확률 0.4)
  + 수학을 좋아하는 학생의 70%가 시험에 합격 P(B|A) = 0.7
  + 질문 : 전체 학생 중, 수학을 좋아하면서 시험에 합격할 비율은?
    

```python
P_A = 0.6
P_B_given_A = 0.7

# 조건부 확률 공식
P_A_and_B = P_A * P_B_given_A
print(P_A_and_B)
```

- 다변량 데이터 분석을 활용할 때도, 조건부 확률의 개념이 들어가지 않을까? 추측

```python
# 결합 확률 정의
from pgmpy.factors.discrete import JointProbabilityDistribution as JPD

# 하나의 사건만 가지고 확률 예감
# 지금은 두개의 사건을 조합해서 확률 값을 구할 수 있다.
# 결합확률밀도 함수

prob = list()
for i in np.arange(1, 4):
    for j in np.arange(1,3):
        prob.append((i+j)/21)
        
fxy = JPD(['X1', 'X2'],[3, 2], prob)
print(fxy)
```

## 결합 확률밀도함수 시각화, 해석

```python
x = np.linspace(0, 1, 100) # 독립적인 사건, 서로 다른 수치형 필드
y = np.linspace(0, 1, 100) # 독립적인 사건, 서로 다른 수치형 필드
X, Y =np.meshgrid(x,y)
X, Y

# 결합 확률 밀도 함수
Z = np.ones_likr(X) #모든 구간에서 밀도 1 # 특정한 시공간이 있는데, 데이터의 밀도는 일정
Z
```

## 상관계수

```python
# 결합확률 정의
prob = np.array([2, 4, 3, 1, 1, 4])/15
fxy = JPD(['X', 'Y'],[2, 3], prob)
print(fxy)
```

```python
# 확률 분포 확인
np.sum(fxy.values)
```

```python
# X 주변 확률 구하기
fx = fxy.marginal_distribution(['X'], inplace=False)
print(fx)
```

```python
# X의 기대값 구하기
x = np.array([1, 2])
EX = np.dot(x,fx.values )
EX.round(3)
```

```python
# X의 제곱 기대값 구하기
EXX = np.dot(x*x, fx.values)
EXX.round(3)
```

```python
# X의 표준편차 구하기
SDX = np.sqrt(EXX-EX**2)
SDX.round(3)
```

```python
# Y의 주변 분포, 기대값, 제곱합, 표준편차

# Y의 주변 분포
fy = fxy.marginal_distribution(['Y'], inplace=False)

# Y의 기대값
y = np.array([1, 2, 3])
EY = np.dot(y,fy.values )

# Y의 제곱 기대값
EYY = np.dot(y*y, fy.values)

# Y의 표준편차
SDY = np.sqrt(EYY-EY**2)
```

```python
# 공분산 구하기
# xy 값 구성
xy = np.outer(x, y).reshape(-1,)
xy
```

```python
# XY 기대값
EXY = np.dot(xy, fxy.values.reshape(-1, ))
EXY.round(3)
```

```python
# 공분산 계산
CovXY = EXY - EX*EY
CovXY.round(3)
```

```python
# 상관계수 값
rho = CovXY / (SDX * SDY)
rho.round(3)
```

```python
# 두 변수 간의 독립성 여부 확인
fxy.check_independence(['X'], ['Y'])
```
